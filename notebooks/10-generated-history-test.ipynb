```python
import sys
sys.path.append('../src')
from loaders import load_frequencies, load_assets, load_procedures, load_service_history
from datetime import datetime
from loaders import DATE_FMT
```

```python
surv_good, _ = load_assets('../data/generated/assets.csv')
valid_asset_ids = {row['asset_id'] for line_num, row in surv_good}
frequencies = load_frequencies('../data/frequencies.csv')
proc_good, _ = load_procedures('../data/generated/procedures.csv', valid_asset_ids, frequencies)
asset_comm = {row['asset_id']: datetime.strptime(row['commissioning_date'], DATE_FMT) for line_num, row in surv_good}
proc_comm = {row['procedure_id']: asset_comm[row['asset_id']] for line_num, row in proc_good}
today = datetime.now()

good, rejected = load_service_history('../data/generated/service_history.csv', proc_comm, today)
print(len(good), 'good rows')
print(len(rejected), 'rejected rows')
for line_num, reason, row in rejected:
    print(line_num, reason)
```
