```python
import sqlite3
import sys
sys.path.append('../src')
from loaders import load_frequencies
```

```python
conn = sqlite3.connect('frequencies.sqlite')
cur = conn.cursor()

cur.executescript('''
DROP TABLE IF EXISTS Frequencies;

CREATE TABLE Frequencies (
    service_frequency  TEXT PRIMARY KEY,
    service_interval_days    INTEGER NOT NULL
);
''')

frequencies = load_frequencies('../data/frequencies.csv')

for freq, days in frequencies.items():

    cur.execute('''INSERT INTO Frequencies (service_frequency, service_interval_days)
        VALUES ( ?,? )''', (freq, days) )


conn.commit()
```

```python
cur.execute('SELECT * FROM Frequencies')
for row in cur.fetchall():
    print(row)
```

```python
try:
    cur.execute('INSERT INTO Frequencies (service_frequency, service_interval_days) VALUES (?, ?)', ('weekly', 7))
except sqlite3.IntegrityError as e:
    print(e)

try:
    cur.execute('INSERT INTO Frequencies (service_frequency, service_interval_days) VALUES (?, ?)', ('daily', None))
except sqlite3.IntegrityError as e:
    print(e)
```
